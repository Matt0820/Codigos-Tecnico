```python
# ============================================
# ANÁLISE SIMPLES DE DADOS NBA
# ============================================
# Este script realiza uma análise básica a partir de um dataset público da NBA.
#
# O objetivo é demonstrar um fluxo simples de análise de dados utilizando Python:
# - leitura de dados via CSV online
# - filtragem de colunas relevantes
# - cálculo de métricas estatísticas
# - visualização com gráfico de barras
#
# Mesmo sendo um código direto, ele representa um pipeline completo:
# dados → processamento → análise → visualização


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Leitura do dataset diretamente de uma URL pública (fonte externa)
tabelaNBA = pd.read_csv('https://raw.githubusercontent.com/MateusVieiraDev/csvdokevin/refs/heads/main/PlayerIndex_nba_stats.csv')

# Seleção de colunas relevantes para análise:
# - nome do jogador
# - time
# - ano de draft
# - pontos (PTS)
tabelReset = tabelaNBA[['PLAYER_FIRST_NAME','TEAM_NAME','DRAFT_YEAR','PTS']]

# Cálculo da média de pontos dos jogadores
mediaTabela = tabelReset['PTS'].mean()

# Identificação do maior valor de pontos no dataset
valorMax = tabelReset['PTS'].nlargest(1)

# Identificação do menor valor de pontos no dataset
valorMin = tabelReset['PTS'].nsmallest(1)

# Exibição das métricas calculadas
print('Media:' , mediaTabela)
print('Min:', valorMin)
print('max',valorMax)

# Preparação dos dados para visualização:
# Remove registros com valores nulos nas colunas utilizadas
grafico = tabelReset[['PTS', 'PLAYER_FIRST_NAME']].dropna(subset=['PLAYER_FIRST_NAME', 'PTS'])

# Conversão explícita dos nomes para string (garantia de consistência)
categoria = grafico['PLAYER_FIRST_NAME'].astype(str)

# Valores que serão representados no gráfico (pontos)
valores = grafico['PTS']

# Configuração do gráfico de barras
plt.figure(figsize=(12,6))

# Criação do gráfico:
# eixo X → nomes dos jogadores
# eixo Y → pontos
plt.bar(categoria, valores, color='skyblue')

# Rotulação dos eixos
plt.xlabel('Jogadores')
plt.ylabel('Pontos')

# Título do gráfico
plt.title('Pontos por Jogador')

# Ajuste da rotação dos nomes para melhor leitura
plt.xticks(rotation=45, ha='right')

# Ajuste automático de layout para evitar sobreposição
plt.tight_layout()

# Exibição do gráfico
plt.show()
```
